```python
import pandas as pd
import numpy as np

df = pd.read_csv("mumbai_pm25_cleaned.csv")
print(df.head())
print(df.shape)
print(df.columns)
print(df.info())
print(df.describe())
```

```python
print("starting date:", df.datetime.min())
print("ending date:", df.datetime.max())
```

## Clean and Validate time series


```python
data = df.copy()
```

```python
data["datetime"] = pd.to_datetime(data["datetime"])

data = data.sort_values("datetime").reset_index(drop=True)

print("Starting time:", data["datetime"].min())
print("Ending time:", data["datetime"].max())
print("Number of observations:", len(data))
```

```python
empty_columns = data.columns[data.isnull().all()]

print("Completely empty columns:")
print(empty_columns)

data = data.drop(columns=empty_columns)
```

```python
print(data.shape)
```

```python
#count constant columns

constant_columns = [
    column for column in data.columns
    if data[column].nunique(dropna=True) <= 1
]

print(constant_columns)
```

```python
data = data.drop(columns=constant_columns)
```

```python
print("duplicate rows:", data.duplicated().sum())
print("duplicate timestamp:", data["datetime"].duplicated().sum())
```

```python
data = data.set_index("datetime")
print("Detected frequency:", pd.infer_freq(data.index))
```

```python
expected_time = pd.date_range(
    start=data.index.min(),
    end=data.index.max(),
    freq="h"
)

missing_timestamps = expected_time.difference(data.index)

print("Expected number of hours:", len(expected_time))
print("Actual number of rows:", len(data))
print("Missing timestamps:", len(missing_timestamps))
print(missing_timestamps[:20])
```

```python
non_negative_columns = [
    "humidity_percent",
    "wind_speed_10m_kmh",
    "wind_gusts_kmh",
    "precipitation_mm",
    "rain_mm",
    "solar_radiation_wm2",
    "direct_radiation_wm2",
    "diffuse_radiation_wm2",
    "cloud_cover_percent",
    "pm2_5_ugm3",
    "pm10_ugm3",
    "co_ugm3",
    "no2_ugm3",
    "so2_ugm3",
    "o3_ugm3",
    "dust_ugm3",
    "aod"
]

for column in non_negative_columns:
    if column in data.columns:
        negative_count = (data[column] < 0).sum()
        print(f"{column}: {negative_count} negative values")
```

```python
checks = {
    "year": data["year"].eq(data.index.year).all(),
    "month": data["month"].eq(data.index.month).all(),
    "day": data["day"].eq(data.index.day).all(),
    "hour": data["hour"].eq(data.index.hour).all(),
    "day_of_week": data["day_of_week"].eq(data.index.dayofweek).all(),
    "quarter": data["quarter"].eq(data.index.quarter).all()
}

print(checks)
```

```python
#remove unnecessary columns
redundant_time_columns = [
    "day",             # Usually no meaningful monthly pattern
    "week_of_year",    # Overlaps with month
    "is_weekend",      # Can be derived from day_of_week
    "quarter",         # Can be derived from month
    "time_of_day",     # Can be derived from hour
    "is_daytime"       # Can be derived from hour
]
categorical_copy_columns = [
    "humidity_category",
    "wind_category"
]
overlapping_weather_columns = [
    "dew_point_c",             # Derived from temperature and humidity
    "rain_mm",                 # Overlaps with precipitation
    "surface_pressure_hpa",    # Overlaps with pressure_msl_hpa
    "direct_radiation_wm2",    # Keep total solar radiation instead
    "diffuse_radiation_wm2",   # Keep total solar radiation instead
    "sunshine_seconds",        # Overlaps with solar radiation
    "cloud_low_percent",       # Keep total cloud cover
    "cloud_mid_percent",       # Keep total cloud cover
    "cloud_high_percent"       # Keep total cloud cover
]
derived_pollution_columns = [
    "pm_ratio",

    "us_aqi_pm25",
    "us_aqi_pm10",
    "us_aqi_no2",
    "us_aqi_o3",
    "us_aqi_co",

    "eu_aqi",
    "eu_aqi_pm25",
    "eu_aqi_pm10"
]
columns_to_remove = (
    redundant_time_columns
    + categorical_copy_columns
    + overlapping_weather_columns
    + derived_pollution_columns
)

print("Number of columns being removed:", len(columns_to_remove))
print(columns_to_remove)
data.drop(
    columns=columns_to_remove,
    inplace=True,
    errors="ignore"
)
print("New shape:", data.shape)
print("\nRemaining columns:")

for column in data.columns:
    print(column)
```

## Visualization


```python
import matplotlib.pyplot as plt

daily_pollution = data[
    ["pm2_5_ugm3", "pm10_ugm3", "no2_ugm3", "o3_ugm3"]
].resample("D").mean()

daily_pollution.plot(
    subplots=True,
    figsize=(15, 12),
    title="Daily Average Pollution Levels"
)

plt.tight_layout()
plt.show()
```

## Exploring pm2.5

```python
print(
    data["pm2_5_ugm3"].describe(
        percentiles=[0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99]
    )
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.hist(
    data["pm2_5_ugm3"],
    bins=60,
    edgecolor="black"
)

plt.title("Distribution of PM2.5")
plt.xlabel("PM2.5 (µg/m³)")
plt.ylabel("Number of observations")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
daily_pm25 = data["pm2_5_ugm3"].resample("D").mean()

plt.figure(figsize=(15, 5))
plt.plot(daily_pm25.index, daily_pm25)

plt.title("Daily Average PM2.5")
plt.xlabel("Date")
plt.ylabel("PM2.5 (µg/m³)")
plt.grid(alpha=0.3)
plt.show()
```

```python
hourly_pattern = data.groupby("hour")["pm2_5_ugm3"].mean()

plt.figure(figsize=(10, 5))
plt.plot(
    hourly_pattern.index,
    hourly_pattern.values,
    marker="o"
)

plt.title("Average PM2.5 by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Average PM2.5 (µg/m³)")
plt.xticks(range(24))
plt.grid(alpha=0.3)
plt.show()

print(hourly_pattern)
```

```python
monthly_pattern = data.groupby("month")["pm2_5_ugm3"].mean()

plt.figure(figsize=(10, 5))
plt.bar(
    monthly_pattern.index,
    monthly_pattern.values,
    edgecolor="black"
)

plt.title("Average PM2.5 by Month")
plt.xlabel("Month")
plt.ylabel("Average PM2.5 (µg/m³)")
plt.xticks(range(1, 13))
plt.grid(axis="y", alpha=0.3)
plt.show()

print(monthly_pattern)
```

```python
weekday_pattern = (
    data.groupby("day_of_week")["pm2_5_ugm3"].mean()
)

weekday_names = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

plt.figure(figsize=(10, 5))

plt.bar(
    weekday_names,
    weekday_pattern.values,
    edgecolor="black"
)

plt.title("Average PM2.5 by Day of Week")
plt.ylabel("Average PM2.5 (µg/m³)")
plt.xticks(rotation=30)
plt.grid(axis="y", alpha=0.3)
plt.show()

print(weekday_pattern)
```

```python
lags_to_check = [1, 2, 3, 6, 12, 24, 48, 168]

autocorrelation_results = {}

for lag in lags_to_check:
    autocorrelation_results[lag] = (
        data["pm2_5_ugm3"].autocorr(lag=lag)
    )

autocorrelation_results = pd.Series(
    autocorrelation_results,
    name="autocorrelation"
)

print(autocorrelation_results)
```

```python
next_hour_pm25 = data["pm2_5_ugm3"].shift(-1)
next_hour_correlations = (
    data.select_dtypes(include="number")
    .corrwith(next_hour_pm25)
    .sort_values(ascending=False)
)

print(next_hour_correlations)
```

```python
next_hour_correlations.drop(
    "pm2_5_ugm3"
).sort_values().plot(
    kind="barh",
    figsize=(10, 8)
)

plt.title("Current Features vs Next-Hour PM2.5")
plt.xlabel("Pearson Correlation")
plt.grid(axis="x", alpha=0.3)
plt.tight_layout()
plt.show()
```
